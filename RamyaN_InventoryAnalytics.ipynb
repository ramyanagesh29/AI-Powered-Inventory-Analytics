````python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["DayOfWeek"] = df["Date"].dt.dayofweek

df[[
    "Date",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "Demand"
]].head(10)
````

````python
features = [
    "Inventory Level",
    "Price",
    "Discount",
    "Promotion",
    "Competitor Pricing",
    "Epidemic",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "Category",
    "Region",
    "Weather Condition",
    "Seasonality"
]

target = "Demand"

X = df[features]
y = df[target]

print("Features:", X.shape)
print("Target:", y.shape)
````

````python
numeric_features = [
    "Inventory Level",
    "Price",
    "Discount",
    "Promotion",
    "Competitor Pricing",
    "Epidemic",
    "Year",
    "Month",
    "Day",
    "DayOfWeek"
]

categorical_features = [
    "Category",
    "Region",
    "Weather Condition",
    "Seasonality"
]
````

````python
train = df[df["Date"] < "2023-10-01"].copy()
test = df[df["Date"] >= "2023-10-01"].copy()

print("Training records:", len(train))
print("Testing records:", len(test))
````

````python
print("Features:", X.shape)
print("Target:", y.shape)

print("Training records:", len(train))
print("Testing records:", len(test))
````

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ],
    remainder="passthrough"
)
````

````python
X_train = train[features]
y_train = train[target]

X_test = test[features]
y_test = test[target]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
````

````python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            RandomForestRegressor(
                n_estimators=100,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
````

````python
print("Training model...")

model.fit(X_train, y_train)

print("Model training completed!")
````

````python
y_pred = model.predict(X_test)

print("Predictions generated successfully!")
print("First 10 predictions:")
print(y_pred[:10])
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)

print("Model Performance")
print("-----------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
````

````python
results = pd.DataFrame({
    "Actual Demand": y_test.values,
    "Predicted Demand": y_pred
})

results.head(10)
````

````python
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.3
)

plt.xlabel("Actual Demand")
plt.ylabel("Predicted Demand")
plt.title("Actual vs Predicted Demand")

plt.tight_layout()
plt.show()
````

````python
rf_model = model.named_steps["regressor"]
encoder = model.named_steps["preprocessor"]

feature_names = encoder.get_feature_names_out()

importances = rf_model.feature_importances_

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
}).sort_values(
    "Importance",
    ascending=False
)

feature_importance.head(15)
````

````python
top_features = feature_importance.head(15).sort_values(
    "Importance"
)

plt.figure(figsize=(9, 6))

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.title("Top Features Used by the Demand Prediction Model")
plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
````

````python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

tuned_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            RandomForestRegressor(
                n_estimators=150,
                max_depth=20,
                min_samples_split=5,
                min_samples_leaf=2,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
````

````python
print("Training tuned model...")

tuned_model.fit(X_train, y_train)

print("Tuned model training completed!")
````

````python
tuned_pred = tuned_model.predict(X_test)

print("Predictions generated!")
````

````python
tuned_mae = mean_absolute_error(y_test, tuned_pred)

tuned_rmse = np.sqrt(
    mean_squared_error(y_test, tuned_pred)
)

tuned_r2 = r2_score(y_test, tuned_pred)

print("Tuned Model Performance")
print("-----------------------")
print("MAE :", tuned_mae)
print("RMSE:", tuned_rmse)
print("R²  :", tuned_r2)
````

````python
comparison = pd.DataFrame({
    "Model": [
        "Baseline Random Forest",
        "Tuned Random Forest"
    ],
    "MAE": [
        mae,
        tuned_mae
    ],
    "RMSE": [
        rmse,
        tuned_rmse
    ],
    "R2": [
        r2,
        tuned_r2
    ]
})

comparison
````

# Business Insights and Recommendations

Based on the exploratory data analysis and demand prediction model, the following
business insights were identified.

## Key Insights

### 1. Overall Inventory Risk
Out of 76,000 records, 10,394 records (13.68%) have demand greater than the
recorded inventory level. This is treated as a potential inventory-risk
indicator rather than a confirmed stock-out.

### 2. Category-Level Risk
Clothing has the highest potential inventory-risk rate at 19.42%, while
Furniture has the lowest rate at 7.84%.

### 3. Regional Risk
North has the highest potential inventory-risk rate among the four regions at
14.60%.

### 4. Product-Level Risk
Product P0008 has the highest potential inventory-risk rate at 28.39%.
Therefore, P0008 should receive closer inventory monitoring.

### 5. Demand by Category
Groceries has the highest average demand at approximately 120.98 units,
while Furniture has the lowest average demand at approximately 73.58 units.

### 6. Promotion and Demand
Records with promotions have an average demand of 123.27 units compared with
95.03 units for records without promotions. This shows an association between
promotion records and higher demand in the dataset.

### 7. Seasonal Demand
Summer has the highest average demand at approximately 112.86 units, while
Spring has the lowest at approximately 97.70 units.

### 8. Product Demand
P0009 has the highest average demand among the analyzed products at
approximately 118.51 units.

### 9. Demand Prediction
The tuned Random Forest model achieved:
- MAE: 20.14
- RMSE: 27.81
- R²: 0.6168

The model captures a substantial portion of demand variation but still has
prediction error and should therefore be used as a supporting planning tool.

# Business Recommendations

1. Prioritize inventory monitoring for products with high potential
   inventory-risk rates, especially P0008.

2. Review inventory planning for the Clothing category because it has the
   highest category-level potential risk rate.

3. Consider promotional periods when planning inventory because promotional
   records show higher average demand in this dataset.

4. Incorporate seasonal demand patterns into inventory planning, particularly
   during periods associated with higher demand such as Summer.

5. Use demand predictions as a supporting tool for inventory planning rather
   than as a replacement for business judgment.

6. Monitor high-demand products such as P0009 closely to reduce the possibility
   of insufficient inventory when demand is high.

## Limitations

- The dataset represents historical observations and may not represent every
  real-world retail environment.
- The condition `Demand > Inventory Level` is used as a potential inventory-risk
  indicator and should not be interpreted as a confirmed stock-out.
- Relationships identified through group-level analysis represent associations
  and do not establish causation.
- The demand prediction model has prediction error and should be used as a
  decision-support tool rather than as a guaranteed forecast.
- The 2024 portion of the dataset contains only January observations, so it
  should not be directly compared with the complete 2022 and 2023 years.
